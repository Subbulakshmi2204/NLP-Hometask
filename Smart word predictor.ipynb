```python
import pandas as pd
import re
from collections import Counter

#Load CSV
data = pd.read_csv("wikitext-2-train.csv")

#take first colummn
text = " ".join(data.iloc[:,0].dropna().astype(str))

#clean text
text = text.lower()
text = re.sub(r"[^a-z0-9\s]"," ",text)

#tokenize
words = text.split()
print("Total words:",len(words))

#unigram
unigram = Counter(words)

#bigram
bigram = Counter(zip(words,words[1:]))

#trigram
trigram = Counter(zip(words,words[1:],words[2:]))

#user input
sentence = input("Enter a sentence: ").lower()
sentence = re.sub(r"[^a-z0-9\s]","",sentence)
w  = sentence.split()
result = Counter()

#trigram prediction
if len(w) >= 2:
    for (a,b,c),count in trigram.items():
        if a == w[-2] and b == w[-1]:
            result[c] += count

#bigram prediction
if not result:
    for (a,b),count in bigram.items():
        if a == w[-1]:
            result[b] += count

#unigram fallback
if not result:
    result = unigram

#Top 5
print("\nTop 5 predicted next words:")
total = sum(result.values())
for i,(word,count) in enumerate(result.most_common(5),1):
    probability = count/total
    print(i,word,"-",round(probability,3))
```
