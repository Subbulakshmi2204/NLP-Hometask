```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

print("Step 1: Loading dataset....")

#1. Load the given IMDb daataset
df = pd.read_csv("IMDB_Dataset_CLEANED.csv")
print("Dataset loaded successfully")

#2. Check missing values and duplicate reviews
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate reviews:",df["review"].duplicated().sum())

#3. Get English stopwords
stop_words = set(stopwords.words("english"))

#4. Create stemmer
stemmer = PorterStemmer()

#5. Function to clean reviews
def clean_review(text):
    text = text.lower() #lowercase
    text = re.sub(r"<.*?>"," ",text) #remove HTML
    text = re.sub(r"http\S+|www\S+"," ",text) #remove URLS
    text = re.sub(r"[^a-z\s]"," ",text) #remove numbers/special characters
    words = text.split() #tokenize
    words = [w for w in words if w not in stop_words]
    words = [stemmer.stem(w) for w in words] #stemming
    return " ".join(words)

#6. Apply cleaning
df["cleaned_review"] = df["review"].apply(clean_review)

#7. Compare original and cleaned reviews
print("\nOriginal Review")
print(df["review"].iloc[0])

print("\nCleaned Review:")
print(df["cleaned_review"].iloc[0])

#8. Display the dataset
print("\nFinal Dataset:")
print(df.head())

#9. Export preprocessed dataset
df.to_csv("IMDB_Preprocessed.csv",index=False)
print("\nPreprocessed dataset saved successfully!")
```
