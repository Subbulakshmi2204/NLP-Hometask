```python
import re

# 1. Load the spelling dataset
file = open("big.txt", "r", encoding="utf-8")

text = file.read().lower()

file.close()

# 2. Build vocabulary of correctly spelled words
vocabulary = set(re.findall(r"[a-z]+", text))

print("Vocabulary size:", len(vocabulary))


# 3. Edit distance function
def edit_distance(a, b):

    dp = [[0] * (len(b) + 1) for i in range(len(a) + 1)]

    for i in range(len(a) + 1):
        dp[i][0] = i

    for j in range(len(b) + 1):
        dp[0][j] = j

    for i in range(1, len(a) + 1):
        for j in range(1, len(b) + 1):

            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],
                    dp[i][j - 1],
                    dp[i - 1][j - 1]
                )

    return dp[-1][-1]


# 4. Accept search query
query = input("\nEnter search query: ")

print("\nOriginal Query:", query)


# 5. Tokenize query
words = re.findall(r"[a-zA-Z]+", query.lower())

corrected_words = []


# 6. Identify incorrect words
for word in words:

    if word in vocabulary:
        corrected_words.append(word)

    else:

        print("\nIncorrect word:", word)

        best_word = word
        best_distance = 100

        # 7. Calculate edit distance
        for candidate in vocabulary:

            # Consider words with similar length
            if abs(len(word) - len(candidate)) <= 2:

                distance = edit_distance(word, candidate)

                if distance < best_distance:
                    best_distance = distance
                    best_word = candidate

        print("Suggested correction:", best_word)
        print("Edit distance:", best_distance)

        corrected_words.append(best_word)


# 8. Display corrected query
corrected_query = " ".join(corrected_words)

print("\nFinal Corrected Query:", corrected_query)
```
